```python
from comparison.source.ctin.ctin_training import ctin_training

for d_model in [32, 64, 128, 256]:

    for batch_size in [32, 64, 128]:

        ctin_training(
            sensor_frequency=100,
            beta=0.1,
            random_seed=42,
            d_model=d_model,
            batch_size=batch_size
        )
```
